## 1. 구글 드라이브 연결

## 2. 경로 설정

```python
PROJECT_DIR = '/content/drive/MyDrive/파이널 프로젝트/파이널_깃허브'
ZIP_DIR = '/content/drive/MyDrive/파이널 프로젝트/images'
ZIP_NAME = 'train_images 회전완료.zip'
```

## 3. 코랩 내부로 복사 + 사진 압축 풀기
드라이브에서 사진을 바로 읽으면 느려서 코랩 내부(`/content/work`)로 옮겨서 작업

```python
import os, glob, shutil, zipfile
WORK = '/content/work'
os.makedirs(WORK, exist_ok=True)
for f in ['01_preprocessing.py', '02_image_features.py']:
    shutil.copy(f'{PROJECT_DIR}/{f}', WORK)
shutil.copytree(f'{PROJECT_DIR}/data', f'{WORK}/data', dirs_exist_ok=True)

os.makedirs(f'{WORK}/data/train_images', exist_ok=True)
with zipfile.ZipFile(f'{ZIP_DIR}/{ZIP_NAME}') as zf:
    zf.extractall(f'{WORK}/data/train_images')
print('사진 수:', len(glob.glob(f'{WORK}/data/train_images/**/*.jpg', recursive=True)))
```
